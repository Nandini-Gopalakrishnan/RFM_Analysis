```python
import pandas as pd
```

```python
df = pd.read_csv("Online_Retail_sales.csv")
print("Number of rows:", df.shape[0])
```

```python
# Step 1: Drop rows with any empty/null values
df = df.dropna()
```

```python
print("Number of rows:", df.shape[0])
```

```python
# Step 2: Remove rows where Quantity or UnitPrice <= 0
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
print("Number of rows:", df.shape[0])
```

```python
# Remove 'StockCode' column
df.drop(columns=['StockCode'], inplace=True)
```

```python
# Save as CSV
df.to_csv("cleaned_sales_data.csv", index=False)
```
